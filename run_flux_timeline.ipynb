## Import relevant python packages

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import flux_timeline as ft
```

## Load relevant files and define constants

scenarios: DRE erupted volumes, eruption and unrest duration
degas_path: degassing calculation

```python
scenarios = pd.read_csv("scenarios.csv", index_col = [0]) # csv for different scenarios
degas_path = pd.read_csv("evo_open_gas.csv") # csv for degassing path
```

```python
tool = 'EVo' # tool used to create the degassing path
constants = {}
constants['crustal_gradient'] = 0.0037 # crustal gradient in km/bar
constants['depth_crust'] = 30. # depth of crust in km
constants['dP_step'] = 0.5 # pressure step in degassing path in bar
constants['density_magma'] = 2737. # melt density in kg/m3
constants['fraction_melt'] = 0.8 # fraction of magma/lava that is melt in a melt+crystal system (i.e., of DRE)
constants['gas_loss_fraction'] = 0.9999 # fraction of gas that is removed at each step in degassing path
```

## Example curves

```python
scenario = "A"
type = 'preeruption'

x = 1000

degas_path = ft.depth_curves(x, type, scenario, degas_path, scenarios, constants, tool)
```

```python
# plotting

fig = make_subplots(rows=1, cols=5, shared_xaxes=False, shared_yaxes=False, 
                    vertical_spacing=0.03, horizontal_spacing=0.05)

r = 1
df = degas_path
c=1
fig.add_trace((go.Scatter(mode="lines", x=df['wCO2g']*(100*1e5), y=df['depth (m)']/1000.,
                          line_color="firebrick", line_dash = "solid", line_width=2)), row = r, col = c)
fig.update_xaxes(range=(0,8),title="CO<sub>2</sub> (\u00D710<sup>-5</sub> wt%)",col=c)
c=2
fig.add_trace((go.Scatter(mode="lines", x=df['wSTg']*100*1e4, y=df['depth (m)']/1000.,
                          line_color="firebrick", line_dash = "solid", line_width=2)), row = r, col = c)
fig.update_xaxes(range=(0,10),title="S<sub>T</sub> (\u00D710<sup>-4</sub> wt%)",side='top',col=c)
c=3
fig.add_trace((go.Scatter(mode="lines", x=df['magma_mass_kg']*1e-4, y=df['depth (m)']/1000.,fill='tozerox',
                          line_color="firebrick", line_dash = "solid", line_width=2)), row = r, col = c)
fig.update_xaxes(range=(0,40),title="magma (\u00D710<sup>4</sub> kg)",col=c)
c=4
fig.add_trace((go.Scatter(mode="lines", x=df['CO2_mass_kg']*10., y=df['depth (m)']/1000.,fill='tozerox',
                          line_color="firebrick", line_dash = "solid", line_width=2)), row = r, col = c)
fig.update_xaxes(range=(0,3),title="CO<sub>2</sub> (\u00D710<sup>-1</sub> kg)",side='top',col=c)
c=5
fig.add_trace((go.Scatter(mode="lines", x=df['ST_mass_kg'], y=df['depth (m)']/1000.,fill='tozerox',
                          line_color="firebrick", line_dash = "solid", line_width=2)), row = r, col = c)
fig.update_xaxes(range=(0,4),title="S<sub>T</sub> (kg)",col=c)
fig.update_yaxes(range=(40,0))
fig.update_yaxes(title="depth (km)", col=1)

fig.update_layout(height=400, width=700, plot_bgcolor='rgb(255,255,255)' , showlegend = False)

fig.update_xaxes(showline=True, linewidth=1, linecolor='black', mirror=True, 
                 ticks="inside", ticklen=5, title_standoff = 0, tickcolor="black",
                 title_font=dict(size=15, family='Helvetica', color='black'), 
                 tickfont=dict(family='Helvetica', color='black', size=12))
fig.update_yaxes(showline=True, linewidth=1, linecolor='black', mirror=True, 
                 ticks="inside", ticklen=5, title_standoff = 0, tickcolor="black",
                title_font=dict(size=15, family='Helvetica', color='black'), 
                 tickfont=dict(family='Helvetica', color='black', size=12))

fig.add_annotation(x=7, y=2,text='A',row=r, col=1,showarrow=False)
fig.add_annotation(x=9, y=38,text='B',row=r, col=2,showarrow=False)
fig.add_annotation(x=36, y=2,text='C',row=r, col=3,showarrow=False)
fig.add_annotation(x=2.7, y=2,text='D',row=r, col=4,showarrow=False)
fig.add_annotation(x=3.5, y=38,text='E',row=r, col=5,showarrow=False)

fig.update_layout(font_family="Helvetica",font_color="black")

fig.show()
#fig.write_image("figure.png")
```

```python
# plotting

rmax = 4
cmax =2

fig = make_subplots(rows=rmax, cols=cmax, shared_xaxes=False, shared_yaxes=False, 
                    vertical_spacing=0.03, horizontal_spacing=0.05)

r = 1
c = 0
for x in ['A','B','C','D','E','F','G','H']:
    if c < cmax:
        c=c+1
    else:
        c=1
        r=r+1
    df = eval(x+'_flux')
    fig.add_trace((go.Scatter(mode="lines", x=df['time wrt eruption start (days)'], y=df['ST flux (ton/day)']*2,
                          line_color="firebrick", line_dash = "solid", line_width=2)), row = r, col = c)
    df = eval(x+'_flux_closed')
    fig.add_trace((go.Scatter(mode="lines", x=df['time (days)'], y=df['ST flux (ton/day)']*2,
                          line_color="steelblue", line_dash = "dash", line_width=3)), row = r, col = c)
    if c == 2:
        fig.update_yaxes(side = 'right', col=c, row = r)
fig.update_xaxes(title="time before eruption (days)",row=rmax)
fig.update_yaxes(title="SO<sub>2</sub> flux (tons/day)")

fig.update_layout(height=1000, width=700, plot_bgcolor='rgb(255,255,255)' , showlegend = False)

fig.update_xaxes(showline=True, linewidth=1, linecolor='black', mirror=True, 
                 ticks="inside", ticklen=5, title_standoff = 0, tickcolor="black",
                 title_font=dict(size=15, family='Helvetica', color='black'), 
                 tickfont=dict(family='Helvetica', color='black', size=12))
fig.update_yaxes(showline=True, linewidth=1, linecolor='black', mirror=True, 
                 ticks="inside", ticklen=5, title_standoff = 0, tickcolor="black",
                title_font=dict(size=15, family='Helvetica', color='black'), 
                 tickfont=dict(family='Helvetica', color='black', size=12))

fig.update_layout(font_family="Helvetica",font_color="black")

fig.show()
#fig.write_image("figure.png")
```
